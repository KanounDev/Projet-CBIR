```python
import os
RAW = "../data/mirflickr/raw"
print(os.listdir(RAW))
```

```python
for name in os.listdir(RAW):
    n = sum(len(files) for _, _, files in os.walk(os.path.join(RAW, name)))
    print(name, n)
```

```python
import re
def numbers(folder, ext):
    out = []
    for root, _, files in os.walk(os.path.join(RAW, folder)):
        out += [int(re.sub(r"\D", "", f)) for f in files if f.endswith(ext)]
    return out

lic = numbers("license", ".txt")
print("license:", min(lic), max(lic), len(lic))

thumbs = []
for t in ["thumbs1-33", "thumbs34-66", "thumbs67-100"]:
    thumbs += numbers(t, ".jpg")
print("thumbs:", min(thumbs), max(thumbs), len(thumbs))

# how many license files have a Photo url (first 5000 files)
count, seen = 0, 0
for root, _, files in os.walk(os.path.join(RAW, "license")):
    for f in files:
        seen += 1
        if "Photo url" in open(os.path.join(root, f), encoding="utf-8", errors="ignore").read():
            count += 1
        if seen == 5000: break
    if seen == 5000: break
print(count, "of", seen, "have a Photo url")
```
